```python
from langchain_openai import ChatOpenAI
from langchain_core.tools import tool
from langchain_community.tools import DuckDuckGoSearchRun
from langchain.messages import HumanMessage
import requests
```

```python
search_tool = DuckDuckGoSearchRun()
search_tool.invoke("top 5 news in bangladesh today")
```

```python
model = ChatOpenAI(model='gpt-4o-mini')
```

```python
model.invoke("hi")
```

```python
from langchain.agents import create_agent
```

```python
agent = create_agent(
    model=model,
    tools=[search_tool],
    system_prompt='Your are an agent who use search tool and find news for today!'
)
```

```python
agent.invoke(input=HumanMessage('today top news in bangladesh'))
```
